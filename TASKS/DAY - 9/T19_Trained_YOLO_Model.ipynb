```python
import cv2
import matplotlib.pyplot as plt

video="runs/detect/predict/Generate_a_realistic_second.avi"

cap=cv2.VideoCapture(video)

total=int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
middle=total//3

cap.set(cv2.CAP_PROP_POS_FRAMES,middle)

ret,frame=cap.read()

if ret:
    h,w,_=frame.shape

    crop_x=int(w*0.10)
    crop_y=int(h*0.10)

    frame=frame[crop_y:h-crop_y,crop_x:w-crop_x]

    frame=cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(6,6))
    plt.imshow(frame)
    plt.axis("off")
    plt.show()

cap.release()
```
